```python
# sigmoid function
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

x_values = np.linspace(-10, 10, 100)
sigmoid_values = sigmoid(x_values)
import matplotlib.pyplot as plt

plt.plot(x_values, sigmoid_values)
plt.title("Sigmoid Function")
plt.xlabel("Input")
plt.ylabel("Output")
plt.grid()
plt.show()
```

```python
# lets say you have very simple function y = 2x + 3 : you can easily do the derivative
def simple_function(x):
    return 2 * x + 3

def derivative_simple_function(x):
    return 2

x = 5
print("Function output:", simple_function(x))
print("Derivative output:", derivative_simple_function(x))
```

```python
# compare array with list for performance
import time
import numpy as np

# create a large list and array
size = 100000000
list_data = list(range(size))
array_data = np.arange(size)

# measure time for list
start = time.time()
list_result = [x * 2 for x in list_data]
print("List processing time:", time.time() - start)

# measure time for array
start = time.time()
array_result = array_data * 2
print("Array processing time:", time.time() - start)
```

```python
# memory usage
import sys

# create a large list and array
size = 1000
list_data = list(range(size))
array_data = np.arange(size)

# measure memory usage for list
print("List memory usage:", sys.getsizeof(list_data))
# in kb
print("List memory usage (KB):", sys.getsizeof(list_data)/1024)

# measure memory usage for array
print("Array memory usage:", array_data.nbytes)
# in kb
print("Array memory usage (KB):", array_data.nbytes/1024)
```

```python
# random selection time comparison
import time
import numpy as np
size = 1000
list_data = list(range(size))
array_data = np.arange(size)
# measure time for random selection in list
start = time.time()
for _ in range(1000):
    _ = np.random.choice(list_data)
print("List random selection time:", time.time() - start)

# measure time for random selection in array
start = time.time()
for _ in range(1000):
    _ = np.random.choice(array_data)
print("Array random selection time:", time.time() - start)
```

```python
# Conclusion: Numpy arrays are more efficient in terms of memory and performance for numerical computations compared to Python lists.
# Additionally, Numpy provides a wide range of mathematical functions and operations that are optimized for performance.
```
