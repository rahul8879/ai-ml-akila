```python
list = [[12,13,14,15],[13,11,13,54],[23,24,25]]

for sblist in list:
    # print(sblist)
    print("Avg", sum(sblist)/len(sblist))
```

```python
def cal_avg_dict(numbers):

    avg = {}
    for index, sblist in enumerate(numbers):
        avg[index] = sum(sblist)/len(sblist)
    return avg

cal_avg_dict([[12,13,14,15],[13,11,13,54],[23,24,25]])
```

```python
def set_fn(num_list=[]):
    if num_list:
        uniq_list = set(num_list)
        return uniq_list
    else:
        print("Empty list provided")

set_fn([111,22,33,44,22,11,33,44,55])
```

```python
def lst_to_dic(num_lst):
    my_dict ={}
    if num_lst:
        for val in num_lst:
            my_dict[val] = val**2
        return my_dict

lst_to_dic([1, 2, 3, 4])
```
