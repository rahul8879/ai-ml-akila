```python
from studentutils import *

def students_record(**students_marks):
    """student course data and generate useful reports."""
    marks = students_marks['marks']
    # print("Original marks", marks)
    students_avg = []
    uniq_subjects = set()
    for student, subjects in marks.items():
        # print(student, subjects)
        
        students_avg.append(get_average_marks(student, subjects))
        uniq_subjects.update(subjects.keys())

    top_student_name(students_avg)
    print("Unique Subjects:", uniq_subjects)
    marks["Nilan"] = {"Math": 100, "Science": 98, "English": 92} 
    print("Updated students list", list(marks.keys()))

students_record(marks = {
    "Akila": {"Math": 88, "Science": 75, "English": 92},
    "Ravi": {"Math": 90, "Science": 89, "English": 95},
    "Jaya": {"Math": 67, "Science": 81, "English": 78}
})
```
