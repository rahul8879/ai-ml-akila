## NumPy Task: Center and scale
You are given feature values.

1) Mean-center the values
2) Scale by dividing by the maximum absolute value
3) Compute the L2 norm of the scaled vector


```python
import numpy as np
from sklearn.preprocessing import MinMaxScaler

np.set_printoptions(precision=2, suppress=True)

x = np.array([3.0, -1.0, 2.0, 6.0, -2.0])

# TODO: mean-center WHY ? To remove the mean and center the data around zero.
centered = x - np.mean(x)
print(centered)

# TODO: scale by max abs value. WHY? To scale features to a comparable range without shifting their center and without destroying sparsity.(-1, 1)
scaled = centered / np.max(np.abs(centered))
# normalized = MinMaxScaler().fit_transform(centered.reshape(-1, 1)).flatten()

print(centered, scaled)

# TODO: compute L2 norm
l2 = np.linalg.norm(scaled)

expected_centered = x - x.mean()
expected_scaled = expected_centered / np.max(np.abs(expected_centered))
expected_l2 = np.linalg.norm(expected_scaled)

assert centered is not None, "centered is None."
assert scaled is not None, "scaled is None."
assert l2 is not None, "l2 is None."
assert np.allclose(centered, expected_centered), f"Expected {expected_centered}, got {centered}"
assert np.allclose(scaled, expected_scaled), f"Expected {expected_scaled}, got {scaled}"
assert abs(l2 - expected_l2) < 1e-12, f"Expected {expected_l2}, got {l2}"
```

## Pandas Task: Replace, compute, and sort
You are given inventory data with missing `stock`.

1) Fill missing `stock` with the column mean
2) Compute inventory value: `stock * cost`
3) Return total inventory value per category sorted descending


```python
import pandas as pd
np.set_printoptions(precision=2, suppress=True)

inv = pd.DataFrame({
    "category": ["Gadgets", "Gadgets", "Tools", "Tools", "Gadgets"],
    "stock": [10, None, 5, 7, None],
    "cost": [4.0, 4.0, 6.0, 6.0, 4.0],
})

# TODO: fill missing stock with mean
inv_clean = inv.copy()
inv_clean["stock"].fillna(np.mean(inv_clean["stock"]).round(1), inplace=True)

# TODO: compute value and group by category
inv_clean["value"] = inv_clean["stock"] * inv_clean["cost"]
value_by_category = inv_clean.groupby("category")["value"].sum()
print(inv_clean, value_by_category)

expected = pd.Series({"Gadgets": 88.0, "Tools": 72.0}) #category
# Gadgets    98.4
# Tools      72.0

assert value_by_category is not None, "value_by_category is None."
assert value_by_category.equals(expected), f"Expected {expected}, got {value_by_category}"
```
