## Linear Regression Scenario: House Pricing
You are helping a small real-estate office estimate house prices.
The target is price in thousands of dollars (`price_k`).


```python
# Feature columns: [size_in_100s_sqft, bedrooms]
X_raw = np.array([
    [1.0, 1],
    [1.5, 1],
    [2.0, 2],
    [2.5, 2],
    [3.0, 3],
])

# price_k follows a clean linear rule (no noise)
price_k = np.array([
    100.0,
    120.0,
    150.0,
    170.0,
    200.0,
])


# Add intercept column
X = np.c_[np.ones(len(X_raw)), X_raw]
# WHY intercept column:  To allow the model to learn a baseline value when all features are zero.
# If you don’t add an intercept column, the model is forced to fit a line/plane that goes through the origin.
# That assumption is almost never correct in real data
```

### Task 1: Fit the model with the normal equation
Compute `weights` using the normal equation: `weights = (X^T X)^-1 X^T y`.
Return a 1D array: `[intercept, size_coef, bedroom_coef]`.


```python
# TODO: replace None with the computed weights
weights = np.linalg.inv(X.T.dot(X)).dot(X.T).dot(price_k)
print(weights)

expected_weights = np.array([50.0, 40.0, 10.0])
assert weights is not None, "weights is None. Did you compute it?"
assert weights.shape == (3,), f"Expected shape (3,), got {weights.shape}"
assert np.allclose(weights, expected_weights, atol=1e-8), f"weights mismatch: {weights}"

# WHY weights: Weights quantify how much each feature influences the prediction.
# Raw features do not explain relationships by themselves.
# The weights learned during training capture these relationships,
# allowing the model to make accurate predictions on new data.
```

### Task 2: Predict a new price
Use your `weights` to predict the price for a 2.2 (hundreds of sqft), 2-bedroom house.
Store the result in `pred_price_k`.


```python
# TODO: replace None with the prediction in thousands of dollars
pred_price_k = weights.dot(np.array([1, 2.2, 2]))  # 1 for intercept, 2.2 for size, 2 for bedrooms
print(pred_price_k)

expected_pred = 158.0
assert pred_price_k is not None, "pred_price_k is None."
assert abs(pred_price_k - expected_pred) < 1e-8, f"Prediction mismatch: {pred_price_k}"
```

### Task 3: Compute training MSE
Compute the mean squared error (MSE) on the training data and store it in `mse`.


```python
# TODO: replace None with the MSE value 
mse = np.mean((price_k - X.dot(weights)) ** 2)
print(mse)

assert mse is not None, "mse is None."
assert abs(mse) < 1e-10, f"MSE should be near zero, got {mse}"

# WHY Compute training MSE: Computing the training MSE provides a quantitative measure of how well the model fits the training data.
# A low training MSE indicates that the model has learned the underlying patterns in the data effectively.
# It helps identify issues like underfitting or overfitting and guides further model improvements.
```

## Logistic Regression Scenario: Loan Approval
A bank uses logistic regression to predict whether a loan is approved.
Features are `[income_score, debt_score]`.
A higher `income_score` helps approval; a higher `debt_score` hurts approval.


```python
X_lr = np.array([
    [1.0, 0.2],
    [0.5, 1.5],
    [2.0, 1.0],
    [1.5, 0.3],
    [0.1, 2.0],
])
y_lr = np.array([1, 0, 1, 1, 0])

w = np.array([1.2, -0.8])
b = -0.1
```

### Task 4: Implement the sigmoid and compute probabilities
Implement `sigmoid(z)` and compute `proba` for each row of `X_lr`.
`proba` should be a 1D array of probabilities.


```python
# TODO: implement sigmoid and compute proba
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

print(sigmoid(X_lr))

proba = sigmoid(X_lr .dot(w) + b)

expected_proba = np.array([0.71909966, 0.33181223, 0.81757448, 0.81153267, 0.17079548])
assert proba is not None, "proba is None."
assert proba.shape == (5,), f"Expected shape (5,), got {proba.shape}"
assert np.allclose(proba, expected_proba, atol=1e-6), f"proba mismatch: {proba}"

# WHY sigmoid: The sigmoid function maps any real-valued number into the (0, 1) interval,
# making it suitable for modeling probabilities in binary classification tasks.
# EX: medical diagnosis (disease vs. no disease), spam detection (spam vs. not spam), etc.
```

### Task 5: Compute log loss
Compute the average log loss and store it in `log_loss_mean`.


```python
# TODO: replace None with the mean log loss
log_loss_mean = -np.mean(y_lr * np.log(proba) + (1 - y_lr) * np.log(1 - proba))

print(log_loss_mean)
expected_log_loss = 0.2660947460290736
assert log_loss_mean is not None, "log_loss_mean is None."
assert abs(log_loss_mean - expected_log_loss) < 1e-9, f"Log loss mismatch: {log_loss_mean}"

# Log loss? it measures how far the predicted probabilities are from the true labels.
# It penalizes confident wrong predictions very heavily
```

### Task 6: Score a new applicant
For an applicant with `[income_score, debt_score] = [1.0, 1.0]`,
compute `prob_new` and `decision` where decision is 1 if prob >= 0.5 else 0.


```python
# TODO: replace None with prob_new and decision

new_X_lr = np.array([[1.0, 1.0]])
prob_new = sigmoid(new_X_lr .dot(w) + b)
decision = 1 if prob_new > 0.5 else 0

expected_prob_new = 0.574442516811659
expected_decision = 1
assert prob_new is not None and decision is not None, "prob_new or decision is None."
assert abs(prob_new - expected_prob_new) < 1e-9, f"prob_new mismatch: {prob_new}"
assert decision == expected_decision, f"decision mismatch: {decision}"
```
