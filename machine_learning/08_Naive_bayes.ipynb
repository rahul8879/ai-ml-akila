```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.linear_model import LogisticRegression
import pandas as pd
import numpy as np
```

```python
data = {
   'reviews': [
       'I love this product',
       'This is the worst purchase I have ever made',
       'Absolutely fantastic! Exceeded my expectations',
       'Not worth the money',
       'I am very satisfied with my purchase',
       'Terrible quality, do not buy this',
       'Great value for the price',
       'I will never buy this again',
       'Highly recommend to everyone',
       'Disappointed with the performance'
   ],
   'sentiments': [
       'positive',
       'negative',
       'positive',
       'negative',
       'positive',
       'negative',
       'positive',
       'negative',
       'positive',
       'negative'
   ]
}
df = pd.DataFrame(data)
df.head(2)
```

```python
vectorizer = CountVectorizer()
X = vectorizer.fit(df['reviews'])
X_transformed = X.transform(df['reviews'])
X_transformed_df = pd.DataFrame(X_transformed.toarray(), columns=vectorizer.get_feature_names_out())
X_transformed_df['sentiments'] = df['sentiments']
X_transformed_df.head(2)
```
