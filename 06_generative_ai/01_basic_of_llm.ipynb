```python
#ChatPromptTemplate ? why to use it 
# The ChatPromptTemplate is a powerful tool for creating dynamic and
#  context-aware prompts for chat-based AI models. By using a template, 
# you can easily customize the input for different topics and languages, 
# ensuring that the AI understands the context and provides relevant responses. 
# This approach not only saves time but also improves the quality of the interactions with the AI.
# Difference between PromptTemplate and ChatPromptTemplate
# The main difference lies in their intended use cases. While PromptTemplate is designed for static prompts,
# ChatPromptTemplate is specifically tailored for dynamic, context-aware conversations.
```

```python
# Basic Chain
from langchain_core.prompts import ChatPromptTemplate
from langchain_openai import ChatOpenAI
from langchain_core.output_parsers import StrOutputParser

chat = ChatOpenAI(temperature=0)
```

```python
prompt = ChatPromptTemplate.from_messages(
    [
        ("system", "You are a helpful AI Tutor."),
        ("human", "Explain me this topic {query} with 3 bullet points")
    ]
)

parser = StrOutputParser()

chain = prompt | chat | parser

chain.invoke({"query":"What is Generative AI?"})
```

```python
from langchain_core.prompts import FewShotChatMessagePromptTemplate

example_data = [{'input': 'Sad', 'output': 'Very Sad'},
                {'input': 'Happy', 'output': 'Very Happy'},
                {'input': 'Test', 'output': 'Very Test'}
               ]

example_prompt = ChatPromptTemplate.from_messages(
    [
        ('user', '{input}'),
        ('ai','{output}')
    ]
)

few_shot = FewShotChatMessagePromptTemplate(example_prompt=example_prompt,
                                 examples=example_data
                                 )


# combine with final question
final_prompt = ChatPromptTemplate.from_messages([
    ('system', 'You are a helpful assistant who gives short responses'),
    ('user','Bad?'),
    few_shot, 
    ('ai', 'Very bad')])
```

```python
chain = final_prompt | model | parser
```

```python
chain.invoke({"input":"Hi there!"})
```
